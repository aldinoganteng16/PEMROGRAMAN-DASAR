```python
# Soal 2: Fungsi dengan parameter Primitif(Matematika)
def hitung_volume_balok(panjang, lebar, tinggi):
    "Menghitung volume balok dari panjang, lebar, dan tinggi."
    return panjang * lebar * tinggi

p = 10
l = 5
t = 3
volume = hitung_volume_balok(p, l, t)

print(f"\n--- HASIL MODIFIKASI ---")
print(f"Panjang: {p}, Lebar: {l}, Tinggi: {t}")
print(f"Volume balok adalah: {volume}")
```

```python
# SOAL 3: Fungsi dengan Parameter Non-Primitif (list)dan Perulangan (for)
def hitung_rata_rata_valid(data_nilai):
    """Menghitung rata-rata dari list angka, hanya menggunakan nilai 0-100."""
    total = 0
    jumlah_nilai_valid = 0

    for nilai in data_nilai:
        if 0 <= nilai <= 100:
            total += nilai
            jumlah_nilai_valid += 1

    if jumlah_nilai_valid > 0:
        return total / jumlah_nilai_valid
    return 0

nilai_praktikum_modifikasi = [85, 78, 92, 40, 65, 105, 90]
print(f"Nilai rata-rata kelas (modifikasi) adalah: {hitung_rata_rata_valid(nilai_praktikum_modifikasi)}")
```

```python
# SOAL 4: Sistem Penentuan Predikat Nilai (Modifikasi)
def tentukan_predikat_aman(skor):
    """Menentukan predikat (A, B, C, D, E) berdasarkan skor dengan pengecekan tipe data."""

    if not isinstance(skor, (int, float)):
        return "ERROR: Input harus berupa angka (int/float)."
    if not 0 <= skor <= 100:
        return "Nilai di luar rentang (0-100)."
    if skor >= 85:
        return "A"
    elif skor >= 70:
        return "B"
    elif skor >= 60:
        return "C"
    elif skor >= 50:
        return "D"
    else:
        return "E"

print(f"Skor 75 mendapat predikat: {tentukan_predikat_aman(75)}")
print(f"Skor 'Delapan Puluh' mendapat predikat: {tentukan_predikat_aman('Delapan Puluh')}")
print(f"Skor 105 mendapat predikat: {tentukan_predikat_aman(105)}")
```

```python
# SOAL 5: Fungsi dengan implementasi Perulangan While
def hitung_faktorial(n):
    """Menghitung faktorial dari bilangan positif n."""

    if n < 0:
        return "Tidak bisa menghitung faktorial dari bilangan negatif"
    elif n == 0:
        return 1

    hasil = 1
    while n > 1:
        hasil *= n
        n -= 1

    return hasil
bilangan_n = 4
print(f"Faktorial dari {bilangan_n} adalah: {hitung_faktorial(bilangan_n)}")
```

```python
# SOAL 6: Analisis Data Suhu Ekstrem (for dan if/elif)
def analisis_suhu(data_suhu, batas_panas=35, batas_dingin=15):
    """Menghitung jumlah hari Panas dan Dingin dari data_suhu"""
    panas_count = 0
    dingin_count = 0

    for suhu in data_suhu:
        if suhu >= batas_panas:
            panas_count += 1
        elif suhu <= batas_dingin:
            dingin_count += 1

    return panas_count, dingin_count

suhu_mingguan = [36, 30, 14, 25, 38, 16, 28]
jumlah_panas, jumlah_dingin = analisis_suhu(suhu_mingguan)
print(f"Jumlah hari Panas (≥{35}°C): {jumlah_panas}")
print(f"Jumlah hari Dingin (≤{15}°C): {jumlah_dingin}")
```
